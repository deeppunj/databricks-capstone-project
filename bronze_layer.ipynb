# Bronze - Layer Raw data Ingestion

```python
# bring the taxi trip dataset 
df_trips = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema","true") \
    .load("/Volumes/taxicatalog/taxi_project_schema/taxivolume/taxi_trip_data_raw.csv")

df_trips.show(5)
```

```python
# Now that the data is loaded, ​the next step is to store it in the bronze layer. 
# We write the data frame in Delta format ​and save it as a managed table inside 
# our schema with the name bronze taxi trips.
df_trips.write.format("delta").saveAsTable("taxicatalog.taxi_project_schema.bronze_taxi_trips")
```

```python
# now let us bring the zone dataset
df_zone = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/taxicatalog/taxi_project_schema/taxivolume/taxi_zone_lookup.csv")

df_zone.show(5)
```

```python
df_zone.write.format("delta").saveAsTable("taxicatalog.taxi_project_schema.bronze_taxi_zones")
```
